## 2. M1-informed data-quality audit and corrections

M1 identified missing names/descriptions, placeholder names, inverted/redundant breeds, unknown medical statuses and suspicious but potentially valid outliers. We preserve records and the target. Unknown medical codes are kept distinct from `No`; we also report the number of records affected by every fix. Outliers are **not removed automatically** because they may be valid animals/listings.

Crucially, this **row-wise, fixed-rule** cleaning does not fit statistical parameters on validation/test data.

```python
import re

PLACEHOLDER_PATTERN = r"\b(?:no\s+name(?:\s+yet)?|noname|un[- ]?named|not\s+named|nameless|unknown)\b"
MEDICAL_COLS = ["Vaccinated", "Dewormed", "Sterilized"]

def quality_audit(df, name):
    text_name = df["Name"].astype("string")
    numeric = ["Age", "Quantity", "Fee", "VideoAmt", "PhotoAmt"]
    result = {
        "dataset": name,
        "rows": int(len(df)),
        "missing_name": int(df["Name"].isna().sum()),
        "placeholder_name": int(text_name.str.contains(PLACEHOLDER_PATTERN, flags=re.IGNORECASE, na=False).sum()),
        "missing_description": int(df["Description"].isna().sum()),
        "inverted_breed": int(((df["Breed1"] == 0) & (df["Breed2"] > 0)).sum()),
        "duplicated_secondary_breed": int(((df["Breed1"] == df["Breed2"]) & (df["Breed2"] > 0)).sum()),
        "color2_gap": int(((df["Color2"] == 0) & (df["Color3"] > 0)).sum()),
        "unknown_medical_rows": int(df[MEDICAL_COLS].eq(3).any(axis=1).sum()),
        "duplicate_pet_ids": int(df["PetID"].duplicated().sum()),
        "identical_except_petid": int(df.drop(columns=["PetID"]).duplicated().sum()),
        "photo_noninteger": int((df["PhotoAmt"].dropna() % 1 != 0).sum()),
    }
    for col in numeric:
        result[f"{col}_min"] = float(df[col].min())
        result[f"{col}_max"] = float(df[col].max())
    return result

def clean_petfinder(df):
    """Deterministic M1 cleaning. Never learns parameters from the dataset."""
    df = df.copy()
    invalid_breed = (df["Breed1"] == 0) & (df["Breed2"] > 0)
    df.loc[invalid_breed, "Breed1"] = df.loc[invalid_breed, "Breed2"]
    df.loc[invalid_breed, "Breed2"] = 0

    same_breed = (df["Breed2"] == df["Breed1"]) & (df["Breed2"] > 0)
    df.loc[same_breed, "Breed2"] = 0

    gap = (df["Color2"] == 0) & (df["Color3"] > 0)
    df.loc[gap, "Color2"] = df.loc[gap, "Color3"]
    df.loc[gap, "Color3"] = 0

    names = df["Name"].astype("string").str.strip().replace("", pd.NA)
    placeholder = names.str.contains(PLACEHOLDER_PATTERN, flags=re.IGNORECASE, na=False)
    df["Name"] = names.mask(placeholder)
    df["Description"] = df["Description"].fillna("").astype(str)

    for col in ("Age", "Fee", "Quantity", "PhotoAmt"):
        df[col] = pd.to_numeric(df[col], errors="raise")
    if not (df["Age"].ge(0).all() and df["Fee"].ge(0).all() and df["Quantity"].gt(0).all()):
        raise ValueError("Unexpected negative Age/Fee or nonpositive Quantity: inspect original records.")
    if df["PhotoAmt"].isna().any() or not np.allclose(df["PhotoAmt"], df["PhotoAmt"].round()):
        raise ValueError("PhotoAmt has missing or fractional values; manual review required.")
    df["PhotoAmt"] = df["PhotoAmt"].astype(int)
    return df

train_clean = clean_petfinder(raw_train)
kaggle_clean = clean_petfinder(raw_kaggle)

audit = pd.DataFrame([
    quality_audit(raw_train, "Labeled / before"),
    quality_audit(train_clean, "Labeled / after"),
    quality_audit(raw_kaggle, "Kaggle test / before"),
    quality_audit(kaggle_clean, "Kaggle test / after"),
]).set_index("dataset")
audit.to_csv(OUT_DIR / "data_quality_audit.csv")
display(audit.T)
assert len(train_clean) == len(raw_train) and len(kaggle_clean) == len(raw_kaggle)
assert raw_train["AdoptionSpeed"].equals(train_clean["AdoptionSpeed"])
assert audit.loc["Labeled / after", "inverted_breed"] == 0
assert audit.loc["Labeled / after", "duplicated_secondary_breed"] == 0
```

```python
# Validity checks against the official breed-label dictionary (audit, not blind correction).
breed_labels = DATA_DIR / "breed_labels.csv"
if breed_labels.exists():
    labels = pd.read_csv(breed_labels)
    breed_to_type = labels.drop_duplicates("BreedID").set_index("BreedID")["Type"]
    breed_type = train_clean["Breed1"].map(breed_to_type)
    incompatible = train_clean[(train_clean["Breed1"] != 0) & breed_type.notna() &
                               (breed_type != train_clean["Type"])]
    print("Primary breeds apparently inconsistent with animal type:", len(incompatible))
    print("These records are flagged, not changed: some breed codes may be ambiguous.")
else:
    print("Breed label file not found; cross-type validation skipped.")

fig, ax = plt.subplots(figsize=(9, 4.5))
missing_before = raw_train.isna().sum().loc[lambda v: v.gt(0)]
if not missing_before.empty:
    missing_before.sort_values().plot(kind="barh", color="#c46b58", ax=ax)
else:
    ax.text(0.5, 0.5, "No actual NaN values", ha="center", transform=ax.transAxes)
ax.set_title("M1 audit: missing values before cleaning")
ax.set_xlabel("Number of records")
fig.tight_layout()
fig.savefig(FIG_DIR / "missing_values.png", dpi=160)
plt.show()
```

## 3. Define an untouched, rescuer-disjoint holdout

**Why `RescuerID` groups?** M1 found different rescuers in the official train and Kaggle test sets. Plain row-wise stratification would place listings from the same rescuer in different internal partitions, potentially overstating generalization. Two nested `StratifiedGroupKFold` partitions give an approximately **60% train / 20% validation / 20% internal test** division. Fractions may differ slightly because all listings from one rescuer must stay together.

The internal test is used **once at the end**, after model and feature decisions are locked using the validation partition.

```python
from sklearn.model_selection import StratifiedGroupKFold

labels_y = train_clean["AdoptionSpeed"].astype(int).to_numpy()
groups = train_clean["RescuerID"].astype(str).to_numpy()
rows = np.arange(len(train_clean))

outer = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEED)
development_idx, holdout_idx = next(outer.split(rows, labels_y, groups))

inner = StratifiedGroupKFold(n_splits=4, shuffle=True, random_state=SEED + 1)
relative_train, relative_val = next(inner.split(
    development_idx, labels_y[development_idx], groups[development_idx]
))
train_idx = development_idx[relative_train]
val_idx = development_idx[relative_val]

def no_group_overlap(a, b):
    return set(groups[a]).isdisjoint(set(groups[b]))

assert len(set(train_idx) | set(val_idx) | set(holdout_idx)) == len(train_clean)
assert no_group_overlap(train_idx, val_idx)
assert no_group_overlap(train_idx, holdout_idx)
assert no_group_overlap(val_idx, holdout_idx)

split_name = np.full(len(train_clean), "", dtype=object)
split_name[train_idx] = "train"
split_name[val_idx] = "validation"
split_name[holdout_idx] = "internal_test"
manifest = train_clean[["PetID", "RescuerID", "AdoptionSpeed"]].copy()
manifest["split"] = split_name
manifest.to_csv(OUT_DIR / "split_manifest.csv", index=False)

split_summary = pd.crosstab(manifest["split"], manifest["AdoptionSpeed"])
split_summary["total"] = split_summary.sum(axis=1)
split_summary["share_%"] = 100 * split_summary["total"] / len(manifest)
display(split_summary.round(2))
print("Distinct rescuers per split:", manifest.groupby("split")["RescuerID"].nunique().to_dict())
```

## 5. Deterministic tabular feature engineering

Derived features: log-scaled age/fees; free listing; named/unnamed status; mixed breed; group listings; fee and photos per pet; text lengths; medical yes/no/**unknown** counts. These are row-wise calculations with no fitted statistics, so they are consistently defined on all partitions.

The medical indicators improve on the earlier `Medical_Score` alone: code `3 = Not sure` is **not silently conflated** with `2 = No`.

```python
def create_tabular_features(df):
    df = df.copy()
    df["Age_Log"] = np.log1p(df["Age"])
    df["Fee_Log"] = np.log1p(df["Fee"])
    df["Is_Free"] = (df["Fee"] == 0).astype("int8")
    df["Has_Name"] = df["Name"].notna().astype("int8")
    df["Is_Mixed"] = ((df["Breed2"] > 0) | (df["Breed1"] == 307)).astype("int8")
    df["Medical_Score"] = df[MEDICAL_COLS].eq(1).sum(axis=1)
    df["Medical_No_Count"] = df[MEDICAL_COLS].eq(2).sum(axis=1)
    df["Medical_Unknown_Count"] = df[MEDICAL_COLS].eq(3).sum(axis=1)
    df["Is_Group"] = (df["Quantity"] > 1).astype("int8")
    df["Fee_Per_Pet"] = df["Fee"] / df["Quantity"]
    df["Photo_Per_Pet"] = df["PhotoAmt"] / df["Quantity"]
    df["Has_Photo"] = (df["PhotoAmt"] > 0).astype("int8")
    df["Has_Description"] = df["Description"].str.len().gt(0).astype("int8")
    df["Desc_Length"] = df["Description"].str.len()
    df["Desc_Word_Count"] = df["Description"].str.split().str.len()
    return df

tab_train = create_tabular_features(train_clean)
tab_kaggle = create_tabular_features(kaggle_clean)

for frame in (tab_train, tab_kaggle):
    assert np.isfinite(frame[["Age_Log", "Fee_Log", "Fee_Per_Pet", "Photo_Per_Pet"]].to_numpy()).all()

engineered_names = sorted(set(tab_train.columns) - set(raw_train.columns))
print(f"Engineered tabular features: {len(engineered_names)}")
print(engineered_names)
```

## 6. Text features: train-fitted TF-IDF → truncated SVD

**What:** TF-IDF encodes important words and short phrases, while Latent Semantic Analysis (truncated SVD) compresses 5,000 sparse coordinates to 10 dense features. **Why:** Descriptions contain valuable information beyond word count. **How:** Fit the vectorizer and SVD **only on internal training text**, then transform validation, holdout and Kaggle test with exactly the same learned objects. English stopwords are used cautiously; descriptions can also contain Malay/Chinese. The explained variance is reported, not assumed to be high.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD

tfidf = TfidfVectorizer(max_features=5000, min_df=2, ngram_range=(1, 2),
                       stop_words="english", sublinear_tf=True, dtype=np.float32)
train_text = train_clean.iloc[train_idx]["Description"]
train_sparse = tfidf.fit_transform(train_text)
assert train_sparse.shape[1] > TEXT_SVD_COMPONENTS

text_svd = TruncatedSVD(n_components=TEXT_SVD_COMPONENTS, random_state=SEED)
text_svd.fit(train_sparse)

def text_features(descriptions):
    sparse = tfidf.transform(descriptions)
    values = text_svd.transform(sparse).astype("float32")
    return pd.DataFrame(values, columns=[f"Desc_SVD_{j}" for j in range(values.shape[1])])

text_train = text_features(train_clean["Description"])
text_kaggle = text_features(kaggle_clean["Description"])
text_explained = float(text_svd.explained_variance_ratio_.sum())
print(f"Vocabulary: {len(tfidf.vocabulary_)}; TF-IDF → {TEXT_SVD_COMPONENTS} components")
print(f"Cumulative SVD explained variance: {text_explained:.2%}")
```

## 7. Sentiment features with explicit missingness flags

The competition provides per-pet JSON files from Google Natural Language. Some are missing. Instead of filling missing scores with zero (which would make *missing* look like *neutral*), preserve `NaN` plus `Has_Sentiment`. LightGBM handles missing values. A count is printed to audit coverage.

```python
def sentiment_features(df, split_name):
    folder = DATA_DIR / f"{split_name}_sentiment"
    records = []
    for pet_id in df["PetID"]:
        path = folder / f"{pet_id}.json"
        score = magnitude = np.nan
        valid = 0
        if path.exists():
            try:
                with path.open(encoding="utf-8") as f:
                    values = json.load(f).get("documentSentiment", {})
                if values.get("score") is not None and values.get("magnitude") is not None:
                    score = float(values["score"])
                    magnitude = float(values["magnitude"])
                    valid = 1
            except (OSError, ValueError, TypeError, json.JSONDecodeError):
                pass
        records.append((score, magnitude, valid))
    features = pd.DataFrame(records, columns=["Sentiment_Score", "Sentiment_Magnitude", "Has_Sentiment"])
    print(f"{split_name}: valid sentiment metadata {features['Has_Sentiment'].sum():,} / {len(features):,}")
    return features

sent_train = sentiment_features(train_clean, "train")
sent_kaggle = sentiment_features(kaggle_clean, "test")
```

## 8. Images: frozen pretrained ResNet50 → train-fitted SVD

**What:** The default listing photo (`PetID-1.jpg`) is encoded by an ImageNet-pretrained, *frozen* ResNet50 into 2,048 dimensions. We reduce this to 15 image features using an SVD fitted **only on internal training embeddings**. **Why:** Photos contain visible characteristics absent from the numeric columns. **How:** Model inference is batched and cached by PetID; missing/corrupt photos receive a zero embedding and a separate indicator. The frozen ImageNet encoder is not trained on the PetFinder labels. **Performance:** GPU strongly recommended; on CPU this section can be slow.

The default photo alone is used for a tractable, reproducible baseline. Multiple-photo aggregation would be a justified future extension.

```python
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision.models import resnet50, ResNet50_Weights
from PIL import Image
from tqdm.auto import tqdm

torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.benchmark = False
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Image encoder device:", device)
if USE_IMAGE_EMBEDDINGS and device.type == "cpu":
    print("WARNING: ~19,000 CPU ResNet50 forward passes can take a long time. Enable a Colab GPU.")

class PetProfileImages(Dataset):
    def __init__(self, pet_ids, folder, transform):
        self.ids = list(map(str, pet_ids))
        self.folder = Path(folder)
        self.transform = transform

    def __len__(self):
        return len(self.ids)

    def __getitem__(self, i):
        path = self.folder / f"{self.ids[i]}-1.jpg"
        if path.is_file():
            try:
                with Image.open(path) as img:
                    return self.transform(img.convert("RGB")), True
            except (OSError, ValueError):
                pass
        return torch.zeros(3, 224, 224, dtype=torch.float32), False

def extract_resnet_embeddings(df, dataset_label, image_dir, encoder, transform):
    cache_path = CACHE_DIR / f"resnet50_{dataset_label}.npz"
    expected_ids = df["PetID"].astype(str).to_numpy()
    if cache_path.exists():
        with np.load(cache_path, allow_pickle=False) as cached:
            if np.array_equal(cached["ids"], expected_ids):
                print(f"Reusing verified {dataset_label} embedding cache.")
                return cached["embeddings"], cached["valid"]

    ds = PetProfileImages(expected_ids, image_dir, transform)
    loader = DataLoader(ds, batch_size=IMAGE_BATCH_SIZE, num_workers=2,
                        shuffle=False, pin_memory=(device.type == "cuda"))
    pieces, valid_pieces = [], []
    with torch.inference_mode():
        for batch, valid in tqdm(loader, desc=f"ResNet50 {dataset_label}"):
            output = encoder(batch.to(device)).float().cpu().numpy()
            valid_numpy = valid.numpy().astype(bool)
            output[~valid_numpy] = 0.0
            pieces.append(output)
            valid_pieces.append(valid_numpy)
    embeddings = np.concatenate(pieces).astype(np.float32)
    valid_mask = np.concatenate(valid_pieces)
    assert embeddings.shape == (len(df), 2048)
    np.savez_compressed(cache_path, ids=expected_ids, embeddings=embeddings, valid=valid_mask)
    print(f"{dataset_label} readable default photos: {valid_mask.sum():,}/{len(df):,}")
    return embeddings, valid_mask

if USE_IMAGE_EMBEDDINGS:
    weights = ResNet50_Weights.DEFAULT
    encoder = resnet50(weights=weights)
    encoder.fc = nn.Identity()
    encoder = encoder.to(device).eval()
    image_transform = weights.transforms()

    image_raw_train, image_valid_train = extract_resnet_embeddings(
        train_clean, "labeled", DATA_DIR / "train_images", encoder, image_transform)
    image_raw_kaggle, image_valid_kaggle = extract_resnet_embeddings(
        kaggle_clean, "kaggle", DATA_DIR / "test_images", encoder, image_transform)

    img_svd = TruncatedSVD(n_components=IMAGE_SVD_COMPONENTS, random_state=SEED)
    img_svd.fit(image_raw_train[train_idx])
    image_cols = [f"Img_Embed_{j}" for j in range(IMAGE_SVD_COMPONENTS)]
    image_train = pd.DataFrame(img_svd.transform(image_raw_train).astype("float32"), columns=image_cols)
    image_kaggle = pd.DataFrame(img_svd.transform(image_raw_kaggle).astype("float32"), columns=image_cols)
    image_train["Has_Readable_Image"] = image_valid_train.astype("int8")
    image_kaggle["Has_Readable_Image"] = image_valid_kaggle.astype("int8")
    image_explained = float(img_svd.explained_variance_ratio_.sum())
    print(f"Image SVD: 2048 → {IMAGE_SVD_COMPONENTS}; cumulative variance {image_explained:.2%}")

    del encoder, image_raw_train, image_raw_kaggle
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
else:
    img_svd = None
    image_explained = None
    image_train = pd.DataFrame({"Has_Readable_Image": np.zeros(len(train_clean), dtype=np.int8)})
    image_kaggle = pd.DataFrame({"Has_Readable_Image": np.zeros(len(kaggle_clean), dtype=np.int8)})

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(range(1, TEXT_SVD_COMPONENTS + 1), np.cumsum(text_svd.explained_variance_ratio_),
        marker="o", label="Text TF-IDF SVD")
if img_svd is not None:
    ax.plot(range(1, IMAGE_SVD_COMPONENTS + 1), np.cumsum(img_svd.explained_variance_ratio_),
            marker="o", label="Image ResNet50 SVD")
ax.set(xlabel="Number of components", ylabel="Cumulative explained variance",
       title="Latent feature compression: measured information retention")
ax.legend()
fig.tight_layout()
fig.savefig(FIG_DIR / "svd_explained_variance.png", dpi=160)
plt.show()
```

## 9. Assemble a consistent, model-ready feature matrix

Each input partition obtains the same columns. `PetID`, `RescuerID`, `Name`, `Description`, `AdoptionSpeed` are never predictors. We explicitly mark **nominal code columns as categories** (numeric IDs are not ordinal). Unknown validation/Kaggle category levels are mapped to missing, which LightGBM can handle. All categorical vocabularies come from the *internal train* only.

```python
from pandas.api.types import CategoricalDtype

IGNORE_COLUMNS = ["PetID", "RescuerID", "Name", "Description", "AdoptionSpeed"]
CATEGORICAL_COLUMNS = ["Type", "Breed1", "Breed2", "Gender", "Color1", "Color2", "Color3",
                       "State", "MaturitySize", "FurLength", "Health", "Vaccinated", "Dewormed", "Sterilized"]

def assemble_features(tab, text, sentiment, image):
    base = tab.drop(columns=IGNORE_COLUMNS, errors="ignore").reset_index(drop=True)
    return pd.concat([base, text.reset_index(drop=True),
                      sentiment.reset_index(drop=True), image.reset_index(drop=True)], axis=1)

features_all = assemble_features(tab_train, text_train, sent_train, image_train)
features_kaggle = assemble_features(tab_kaggle, text_kaggle, sent_kaggle, image_kaggle)
assert features_all.columns.tolist() == features_kaggle.columns.tolist()
assert features_all.columns.is_unique

category_levels = {}
for col in CATEGORICAL_COLUMNS:
    values = sorted(features_all.iloc[train_idx][col].dropna().unique().tolist())
    category_levels[col] = [int(v) for v in values]
    dtype = CategoricalDtype(categories=category_levels[col], ordered=False)
    features_all[col] = features_all[col].astype(dtype)
    features_kaggle[col] = features_kaggle[col].astype(dtype)

def verify_features(frame):
    numbers = frame.select_dtypes(include=[np.number])
    if np.isinf(numbers.to_numpy(dtype=float)).any():
        raise ValueError("Infinite values found in feature matrix")
    assert "AdoptionSpeed" not in frame and "PetID" not in frame and "RescuerID" not in frame

verify_features(features_all)
verify_features(features_kaggle)
X_train = features_all.iloc[train_idx].copy().reset_index(drop=True)
X_val = features_all.iloc[val_idx].copy().reset_index(drop=True)
X_holdout = features_all.iloc[holdout_idx].copy().reset_index(drop=True)
X_kaggle = features_kaggle.copy().reset_index(drop=True)
y_train = train_clean.iloc[train_idx]["AdoptionSpeed"].astype(int).reset_index(drop=True)
y_val = train_clean.iloc[val_idx]["AdoptionSpeed"].astype(int).reset_index(drop=True)
y_holdout = train_clean.iloc[holdout_idx]["AdoptionSpeed"].astype(int).reset_index(drop=True)

assert X_train.columns.tolist() == X_val.columns.tolist() == X_holdout.columns.tolist() == X_kaggle.columns.tolist()
assert X_train.shape[0] == len(y_train) and X_holdout.shape[0] == len(y_holdout)
print(f"All features: {X_train.shape[1]} | Training listings: {len(X_train):,} | Kaggle test: {len(X_kaggle):,}")
print("Categories:", CATEGORICAL_COLUMNS)
print("Original/engineered tabular, sentiment, TF-IDF/SVD and image embeddings share one schema.")
```

## 11. Feature selection: LightGBM + multiclass SHAP (training data only)

**What:** Train a multiclass LightGBM on **internal training listings** and compute global mean absolute SHAP magnitudes on a stratified *training* sample. **Why:** SHAP estimates feature contributions and provides a principled, repeatable ranking. **How:** Select the `TOP_K_FEATURES` highest-ranked predictors (default 30); no internal-test observations or labels enter this decision.

The held-out validation set is used **only to compare** the full-feature model with a model retrained on the selected features. We evaluate the validation-approved winner once on the untouched internal test. We use QWK, macro-F1, balanced accuracy and accuracy; the Kaggle task is ordinal, so QWK is particularly relevant.

```python
import lightgbm as lgb
import shap
from sklearn.model_selection import train_test_split
from sklearn.metrics import (cohen_kappa_score, f1_score, balanced_accuracy_score,
                             accuracy_score, classification_report, confusion_matrix)

def new_classifier():
    return lgb.LGBMClassifier(
        objective="multiclass", num_class=5, n_estimators=300, learning_rate=0.05,
        num_leaves=31, max_depth=6, min_child_samples=25,
        colsample_bytree=0.9, subsample=0.9, subsample_freq=1,
        class_weight="balanced", random_state=SEED, n_jobs=-1, verbosity=-1
    )

def fit_classifier(train_frame, labels, columns):
    model = new_classifier()
    categories = [c for c in CATEGORICAL_COLUMNS if c in columns]
    model.fit(train_frame[list(columns)], labels, categorical_feature=categories)
    return model

full_columns = X_train.columns.tolist()
full_model = fit_classifier(X_train, y_train, full_columns)

# A stratified training-only subset reduces SHAP computation time and avoids val/test influence.
n_shap = min(1500, len(X_train))
if n_shap < len(X_train):
    _, shap_idx = train_test_split(np.arange(len(X_train)), test_size=n_shap,
                                  stratify=y_train, random_state=SEED)
else:
    shap_idx = np.arange(len(X_train))
X_shap = X_train.iloc[shap_idx].copy()

explainer = shap.TreeExplainer(full_model)
raw_shap = explainer.shap_values(X_shap)
if isinstance(raw_shap, list):
    values = np.stack([np.asarray(v) for v in raw_shap], axis=-1)
else:
    values = np.asarray(raw_shap)
    if values.ndim == 2:
        values = values[:, :, None]
    if values.ndim == 3 and values.shape[1] != len(full_columns) and values.shape[2] == len(full_columns):
        values = np.transpose(values, (1, 2, 0))
assert values.ndim == 3 and values.shape[:2] == (len(X_shap), len(full_columns)), values.shape

importance = np.abs(values).mean(axis=(0, 2))
ranking = pd.DataFrame({"feature": full_columns, "mean_abs_shap": importance})
ranking = ranking.sort_values("mean_abs_shap", ascending=False, kind="mergesort").reset_index(drop=True)
ranking["rank"] = np.arange(1, len(ranking) + 1)
selected_columns = ranking.head(min(TOP_K_FEATURES, len(ranking)))["feature"].tolist()
ranking["selected"] = ranking["feature"].isin(selected_columns)
ranking.to_csv(OUT_DIR / "feature_ranking_shap.csv", index=False)
display(ranking.head(30))

fig, ax = plt.subplots(figsize=(10, 8))
top = ranking.head(20).iloc[::-1]
ax.barh(top["feature"], top["mean_abs_shap"], color="#37849a")
ax.set_xlabel("Mean absolute multiclass SHAP value")
ax.set_title("Global feature importance — training sample only")
fig.tight_layout()
fig.savefig(FIG_DIR / "shap_global_importance.png", dpi=160)
plt.show()

# Inspect one specific class while selecting with global importance across *all* classes.
try:
    shap.summary_plot(values[:, :, 0], X_shap, max_display=18, show=False)
    plt.title("SHAP contributions for AdoptionSpeed = 0 (training only)")
    plt.tight_layout()
    plt.savefig(FIG_DIR / "shap_class0_summary.png", dpi=160, bbox_inches="tight")
    plt.show()
except (TypeError, ValueError) as error:
    print("SHAP beeswarm skipped (package version or categorical rendering):", error)

print(f"Automatically selected {len(selected_columns)} features:")
print(selected_columns)
```

## 12. Validate the selected feature set without touching the internal test

Model hyperparameters are held fixed. The decision between the full set and the SHAP-selected set uses **validation QWK** only. The holdout remains sealed until the next section. The selection might not outperform all features: documenting this is part of responsible feature analysis.

```python
def classification_metrics(y_true, y_pred):
    return {
        "QWK": float(cohen_kappa_score(y_true, y_pred, weights="quadratic", labels=[0, 1, 2, 3, 4])),
        "macro_F1": float(f1_score(y_true, y_pred, average="macro", labels=[0, 1, 2, 3, 4], zero_division=0)),
        "balanced_accuracy": float(balanced_accuracy_score(y_true, y_pred)),
        "accuracy": float(accuracy_score(y_true, y_pred)),
    }

selected_model = fit_classifier(X_train, y_train, selected_columns)
full_val_pred = full_model.predict(X_val[full_columns])
selected_val_pred = selected_model.predict(X_val[selected_columns])

validation_comparison = pd.DataFrame([
    {"variant": "all_features", "n_features": len(full_columns), **classification_metrics(y_val, full_val_pred)},
    {"variant": "shap_top_k", "n_features": len(selected_columns), **classification_metrics(y_val, selected_val_pred)},
]).set_index("variant")
validation_comparison.to_csv(OUT_DIR / "validation_feature_comparison.csv")
display(validation_comparison.round(4))

use_selected = validation_comparison.loc["shap_top_k", "QWK"] >= validation_comparison.loc["all_features", "QWK"]
winner_name = "shap_top_k" if use_selected else "all_features"
chosen_columns = selected_columns if use_selected else full_columns
chosen_model = selected_model if use_selected else full_model
print(f"Chosen by validation QWK: {winner_name} ({len(chosen_columns)} features)")
```

## 14. Feature-space diagnostics: PCA and t-SNE

PCA provides a linear 2D projection and explained-variance percentages; t-SNE provides a local-neighborhood visualization. The selected feature matrix mixes continuous and nominal data, so we use **train-fitted numeric median imputation + standardization and categorical one-hot encoding** for visualization. This preprocessing is **not used for LightGBM** (which handles categories and missing values directly).

We fit PCA on training data only and run t-SNE on an explicitly capped, stratified sample of training listings to keep a fresh Colab execution practical. Neither plot is a quantitative proof of class separability.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

visual_cat = [c for c in CATEGORICAL_COLUMNS if c in chosen_columns]
visual_num = [c for c in chosen_columns if c not in visual_cat]
visual_transformers = []
if visual_num:
    visual_transformers.append(("numeric", Pipeline([
        ("impute", SimpleImputer(strategy="median", keep_empty_features=True)),
        ("scale", StandardScaler()),
    ]), visual_num))
if visual_cat:
    visual_transformers.append(("categorical", OneHotEncoder(
        handle_unknown="ignore", sparse_output=False), visual_cat))

visual_preprocessor = ColumnTransformer(visual_transformers, remainder="drop", sparse_threshold=0)
Z_train = visual_preprocessor.fit_transform(X_train[chosen_columns])
Z_train = np.asarray(Z_train, dtype=np.float32)
assert np.isfinite(Z_train).all()
print("Visualization-only one-hot/standardized feature space:", Z_train.shape)

pca = PCA(n_components=2, svd_solver="randomized", random_state=SEED)
P_train = pca.fit_transform(Z_train)

fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(x=P_train[:, 0], y=P_train[:, 1], hue=y_train.astype(str),
                palette="viridis", alpha=0.45, s=14, linewidth=0, ax=ax)
ax.set(xlabel=f"PC1 ({pca.explained_variance_ratio_[0]:.1%})",
       ylabel=f"PC2 ({pca.explained_variance_ratio_[1]:.1%})",
       title="PCA: training feature space (no test leakage)")
ax.legend(title="AdoptionSpeed", bbox_to_anchor=(1.02, 1), loc="upper left")
fig.tight_layout()
fig.savefig(FIG_DIR / "pca_feature_space.png", dpi=160)
plt.show()

if len(Z_train) > TSNE_SAMPLE_SIZE:
    _, plot_idx = train_test_split(np.arange(len(Z_train)), test_size=TSNE_SAMPLE_SIZE,
                                  random_state=SEED, stratify=y_train)
else:
    plot_idx = np.arange(len(Z_train))
Z_sample = Z_train[plot_idx]
plot_labels = y_train.iloc[plot_idx].astype(str)
tsne = TSNE(n_components=2, perplexity=min(30, max(5, (len(Z_sample) - 1) // 3)),
            init="pca", learning_rate="auto", random_state=SEED)
T_sample = tsne.fit_transform(Z_sample)

fig, ax = plt.subplots(figsize=(8, 6))
sns.scatterplot(x=T_sample[:, 0], y=T_sample[:, 1], hue=plot_labels,
                palette="viridis", alpha=0.55, s=16, linewidth=0, ax=ax)
ax.set(xlabel="t-SNE 1", ylabel="t-SNE 2",
       title=f"t-SNE: stratified training subset (n={len(plot_idx):,})")
ax.legend(title="AdoptionSpeed", bbox_to_anchor=(1.02, 1), loc="upper left")
fig.tight_layout()
fig.savefig(FIG_DIR / "tsne_feature_space.png", dpi=160)
plt.show()

print("PCA variance explained by first two components:",
      round(100 * float(pca.explained_variance_ratio_.sum()), 2), "%")
```

```python
from reportlab.platypus import (SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle,
                                 Image as RLImage, PageBreak, KeepTogether)
from reportlab.lib import colors
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.pagesizes import A4
from pypdf import PdfReader

REPORT_PATH = OUT_DIR / "M2_Data_Preprocessing_and_Feature_Analysis.pdf"
styles = getSampleStyleSheet()
styles.add(ParagraphStyle(name="M2Title", parent=styles["Title"], fontName="Helvetica-Bold",
                          fontSize=18, leading=22, textColor=colors.HexColor("#193e55"), spaceAfter=10))
styles.add(ParagraphStyle(name="M2Heading", parent=styles["Heading2"], fontName="Helvetica-Bold",
                          fontSize=12, leading=15, textColor=colors.HexColor("#1e4b61"), spaceBefore=10, spaceAfter=6))
styles.add(ParagraphStyle(name="M2Body", parent=styles["Normal"], fontSize=8.6,
                          leading=12.2, spaceAfter=6))
styles.add(ParagraphStyle(name="M2Small", parent=styles["Normal"], fontSize=7.4,
                          leading=10.5, spaceAfter=4))

def para(text, style="M2Body"):
    return Paragraph(text, styles[style])

def pdf_table(data, col_widths=None):
    t = Table([[str(x) for x in row] for row in data], colWidths=col_widths, repeatRows=1,
              hAlign="LEFT")
    t.setStyle(TableStyle([
        ("BACKGROUND", (0, 0), (-1, 0), colors.HexColor("#264f67")),
        ("TEXTCOLOR", (0, 0), (-1, 0), colors.white),
        ("ROWBACKGROUNDS", (0, 1), (-1, -1), [colors.white, colors.HexColor("#eff4f6")]),
        ("FONTNAME", (0, 0), (-1, 0), "Helvetica-Bold"),
        ("FONTNAME", (0, 1), (-1, -1), "Helvetica"),
        ("FONTSIZE", (0, 0), (-1, -1), 7.5),
        ("LEADING", (0, 0), (-1, -1), 10.0),
        ("BOTTOMPADDING", (0, 0), (-1, -1), 5),
        ("TOPPADDING", (0, 0), (-1, -1), 5),
        ("VALIGN", (0, 0), (-1, -1), "MIDDLE"),
        ("LINEBELOW", (0, 0), (-1, 0), 0.6, colors.HexColor("#18394d")),
    ]))
    return t

def fit_figure(name, width=460, height=230):
    path = FIG_DIR / name
    with Image.open(path) as im:
        original_w, original_h = im.size
    ratio = min(width / original_w, height / original_h)
    return RLImage(str(path), width=original_w * ratio, height=original_h * ratio)

pdf = SimpleDocTemplate(str(REPORT_PATH), pagesize=A4,
                        leftMargin=43, rightMargin=43, topMargin=42, bottomMargin=42,
                        title="PetFinder M2 - Data Preprocessing and Feature Analysis",
                        author="Team A")

story = []

# PAGE 1: objective and quality.
story += [para("PetFinder M2 | Preprocessing and Feature Analysis", "M2Title"),
          para("Team A | Reproducible Google Colab workflow | Milestone 2", "M2Small"),
          para("<b>Objective.</b> Predict the five ordered AdoptionSpeed categories from tabular, text "
               "and image information. The labeled Kaggle dataset has 14,993 profiles; the official "
               "3,972-row Kaggle test has no outcomes. We extend the M1 audit and avoid all supervised "
               "evaluation on the Kaggle test."),
          para("1. Data quality: what, why and how", "M2Heading"),
          para("M1 found missing descriptions/names, placeholder names, duplicated or inverted breed "
               "codes and heavily skewed numeric variables. Deterministic cleaning swaps valid secondary "
               "breeds into missing primary slots, removes duplicate secondary breeds, closes Color2/Color3 "
               "gaps and normalizes empty names/descriptions. Plausible numerical outliers are retained; "
               "unknown medical statuses stay explicit categories. Neither rows nor target labels are removed."),
          pdf_table([
              ["Labeled-train audit", "Before", "After"],
              ["Missing Name", audit.loc["Labeled / before", "missing_name"],
               audit.loc["Labeled / after", "missing_name"]],
              ["Placeholder names", audit.loc["Labeled / before", "placeholder_name"],
               audit.loc["Labeled / after", "placeholder_name"]],
              ["Missing Description", audit.loc["Labeled / before", "missing_description"],
               audit.loc["Labeled / after", "missing_description"]],
              ["Inverted Breed1", audit.loc["Labeled / before", "inverted_breed"],
               audit.loc["Labeled / after", "inverted_breed"]],
              ["Repeated Breed2", audit.loc["Labeled / before", "duplicated_secondary_breed"],
               audit.loc["Labeled / after", "duplicated_secondary_breed"]],
              ["Unknown medical (kept)", audit.loc["Labeled / before", "unknown_medical_rows"],
               audit.loc["Labeled / after", "unknown_medical_rows"]],
          ], col_widths=[235, 115, 115]),
          Spacer(1, 10), fit_figure("missing_values.png", 440, 165), PageBreak()]

# PAGE 2: group-aware splits, feature engineering and feature variance.
story += [para("2. Leakage-safe data splits and transformations", "M2Title"),
          para("<b>Split.</b> Two nested StratifiedGroupKFold partitions group all profiles by RescuerID "
               "and approximately preserve the five outcome proportions. A rescuer cannot occur in "
               "multiple internal partitions. The final holdout stays untouched during feature/model choice."),
          pdf_table([["Partition", "Rows", "Share"]] + [
              [key, int((split_name == key).sum()),
               f"{100 * (split_name == key).mean():.1f}%"]
              for key in ("train", "validation", "internal_test")
          ], col_widths=[235, 115, 115]),
          Spacer(1, 8),
          para("<b>Tabular.</b> log1p(Age/Fee) reduces right skew; free/named/mixed/group indicators, "
               "per-pet counts and description lengths capture additional signals. Medical Yes/No/Not sure "
               "counts avoid treating uncertainty as a negative test result."),
          para("<b>Text.</b> Training-only TF-IDF (5,000 features, word unigrams/bigrams) and "
               f"{TEXT_SVD_COMPONENTS}-component SVD summarize descriptions "
               f"(explained variance = {text_explained:.1%})."),
          para("<b>Images.</b> Frozen ImageNet ResNet50 extracts 2,048 profile-photo coordinates. "
               + (f"Training-only {IMAGE_SVD_COMPONENTS}-component SVD retains "
                  f"{image_explained:.1%} of their sample variance."
                  if img_svd is not None else "Image extraction was disabled for this run.")),
          para("<b>Sentiment.</b> Official JSON sentiment score/magnitude use explicit availability "
               "indicators; missing files are NaN, not false neutral scores. Category vocabularies are "
               "learned from training rows only."),
          fit_figure("svd_explained_variance.png", 440, 225),
          PageBreak()]

# PAGE 3: SHAP and model comparison.
story += [para("3. Feature selection and validation", "M2Title"),
          para(f"A five-class LightGBM with fixed parameters was fitted on internal train rows. "
               f"Mean absolute multiclass TreeSHAP over {len(X_shap):,} stratified training listings "
               f"ranked all {len(full_columns)} predictors. The top {len(selected_columns)} features "
               "were selected automatically; the internal holdout was not consulted."),
          fit_figure("shap_global_importance.png", 465, 340),
          para("Full vs SHAP-selected features: identical model settings and validation-only comparison.", "M2Small"),
          pdf_table([["Variant", "Features", "QWK", "Macro-F1"]] + [
              [key, int(row["n_features"]), f"{row['QWK']:.4f}", f"{row['macro_F1']:.4f}"]
              for key, row in validation_comparison.iterrows()
          ], col_widths=[170, 85, 105, 105]),
          Spacer(1, 8),
          para(f"<b>Validation decision:</b> {winner_name} ({len(chosen_columns)} features). "
               "A feature set is preferred only if validation QWK supports it. SHAP importance "
               "describes a fitted model, not causal importance; correlated features can share credit."),
          PageBreak()]

# PAGE 4: exploratory projections.
story += [para("4. Feature-space exploration", "M2Title"),
          para("Numeric variables are median-imputed and standardized; nominal variables use "
               "training-fitted one-hot encoding. These steps are for visual diagnostics only, not "
               "the LightGBM predictor. PCA is fitted only on internal train; t-SNE uses a capped "
               "stratified training subset."),
          pdf_table([["Diagnostic", "Reported value"],
                     ["PCA PC1 + PC2 explained variance",
                      f"{float(pca.explained_variance_ratio_.sum()):.2%}"],
                     ["t-SNE training sample", f"{len(plot_idx):,} listings"]],
                    col_widths=[290, 175]),
          Spacer(1, 8),
          Table([[fit_figure("pca_feature_space.png", 242, 202),
                  fit_figure("tsne_feature_space.png", 242, 202)]], colWidths=[247, 247]),
          Spacer(1, 8),
          fit_figure("feature_correlations.png", 405, 290),
          para("Interpretation: local visual overlap is expected, as many demographic and listing "
               "features only weakly explain adoption. t-SNE distances and apparent clusters "
               "must not be treated as quantitative performance or probabilities.", "M2Small"),
          PageBreak()]

# PAGE 5: final untouched-test metrics and reproducibility.
story += [para("5. Locked holdout and reproducibility", "M2Title"),
          para("Only after validation chose the feature set did we evaluate the internal rescuer-disjoint "
               "test once. Quadratic weighted kappa (QWK) penalizes larger mistakes more strongly; "
               "macro-F1 gives the rare same-day adoption class equal weight."),
          pdf_table([["Internal holdout", "Value"],
                     ["Chosen variant", winner_name],
                     ["Quadratic weighted kappa", f"{holdout_metrics.loc[0, 'QWK']:.4f}"],
                     ["Macro-F1", f"{holdout_metrics.loc[0, 'macro_F1']:.4f}"],
                     ["Balanced accuracy", f"{holdout_metrics.loc[0, 'balanced_accuracy']:.4f}"],
                     ["Accuracy", f"{holdout_metrics.loc[0, 'accuracy']:.4f}"]],
                    col_widths=[290, 175]),
          Spacer(1, 6), fit_figure("holdout_confusion_matrix.png", 360, 285),
          para("<b>Reproduce:</b> Run the supplied Colab from top to bottom with Kaggle access "
               f"and SEED={SEED}. The output includes exact split IDs, package versions, data-quality "
               "audit, fitted transformers/model, SHAP ranking, aligned train/validation/holdout/Kaggle "
               "feature CSVs, the competition-format submission and all figures."),
          para("<b>Limitations:</b> A frozen ResNet50 uses only the default first image; the official "
               "Kaggle test has no available labels; SHAP estimates are model-dependent. Group-based "
               "splits approximate, rather than guarantee, identical class ratios."),
          para("<b>Sources:</b> Team A (M1 Data Exploration); PetFinder.my Adoption Prediction, "
               "https://www.kaggle.com/competitions/petfinder-adoption-prediction/data", "M2Small")]

def add_pdf_page_number(canvas, document):
    canvas.setFont("Helvetica", 8)
    canvas.setFillColor(colors.HexColor("#5b6c75"))
    canvas.drawRightString(A4[0] - 42, 24, f"Team A | M2 | Page {document.page}")

pdf.build(story, onFirstPage=add_pdf_page_number, onLaterPages=add_pdf_page_number)
report_pages = len(PdfReader(str(REPORT_PATH)).pages)
if report_pages > 5:
    raise RuntimeError(f"PDF is {report_pages} pages; adjust plot sizes/content to meet the 5-page maximum.")
print(f"PDF documentation generated: {REPORT_PATH} ({report_pages} pages)")
```
